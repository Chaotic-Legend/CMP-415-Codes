In this homework assignment, you will implement the normal equation for simple linear regression using a non-matrix format and verify that the parameter values indeed make the partial derivatives of the loss function equal to zero.

a. Load the advertising dataset from this week's class notebook. Only keep the "TV" column as an input feature and the "sales" column as an output feature. Display the first 5 rows of the data frame.

```python
import pandas as pd
import numpy as np
# Load the advertising dataset.
url = "https://www.statlearning.com/s/Advertising.csv"
Advertising = pd.read_csv(url)

# Only keep the "TV" column and the "sales" column.
Advertising = Advertising[["TV", "sales"]]

# Display the first five rows of the data frame.
print("=== Display the Advertising Dataset ===")
Advertising.head()
```

b. Apply the following formulas to calculate the optimal parameter values:

$$
\bar{x} = \frac{1}{m}\sum_{i=0}^{m-1} x^{(i)}
$$

$$
\bar{y} = \frac{1}{m}\sum_{i=0}^{m-1} y^{(i)}
$$

$$
\theta_1 = \frac{\sum_{i=0}^{m-1}(x^{(i)} - \bar{x})(y^{(i)} - \bar{y})}{\sum_{i=0}^{m-1}(x^{(i)} - \bar{x})^2}
$$

$$
\theta_0 = \bar{y} - \theta_1\bar{x}
$$

Display the value of $\bar{x}, \bar{y}, \theta_0, \theta_1$.

```python
# Extract the values.
x = Advertising["TV"].values
y = Advertising["sales"].values
m = len(x)

# Compute the means.
x_bar = sum(x) / m
y_bar = sum(y) / m

# Compute the theta_1 value.
numerator = sum((x[i] - x_bar) * (y[i] - y_bar) for i in range(m))
denominator = sum((x[i] - x_bar) ** 2 for i in range(m))
theta_1 = numerator / denominator

# Compute the theta_0 value.
theta_0 = y_bar - theta_1 * x_bar

print("=== Normal Equation Parameter Computation ===")
print("x̄ =", x_bar)
print("\nȳ =", y_bar)
print("\nθ₀ =", theta_0)
print("\nθ₁ =", theta_1)
```

c. The partial derivative of the MSE cost function are:

$$\begin{align}
\frac{\partial MSE}{\partial \theta_0} =& -2 \sum_{i=0}^{m-1}\big(\theta_0 + \theta_1x^{(i)} - y^{(i)}\big)\\
\frac{\partial MSE}{\partial \theta_1} =& -2 \sum_{i=0}^{m-1} x_i\big(\theta_0 + \theta_1x^{(i)} - y^{(i)}\big).
\end{align}$$

Display the value of these partial derivatives for $\theta_0, \theta_1$ calculated above.

```python
# Compute the value of these partial derivatives.
partial_theta_0 = -2 * sum((theta_0 + theta_1 * x[i] - y[i]) for i in range(m))
partial_theta_1 = -2 * sum(x[i] * (theta_0 + theta_1 * x[i] - y[i]) for i in range(m))
print("=== Verification of MSE Partial Derivatives ===")
print("∂MSE/∂θ₀ =", partial_theta_0)
print("\n∂MSE/∂θ₁ =", partial_theta_1)
```
