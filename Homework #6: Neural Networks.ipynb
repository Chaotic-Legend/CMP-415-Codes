# 1. Network Parameters

**Inputs:**

$$
x_1 = 1,\quad x_2 = 1
$$

**Hidden layer weights and biases:**

- To $H_1$:

$$
w_{1,1} = 4.83,\; w_{2,1} = 4.83,\; b_1 = -2.82
$$

- To $H_2$:

$$
w_{1,2} = 4.54,\; w_{2,2} = 4.60,\; b_2 = -2.74
$$

**Output layer weights and bias:**

$$
w_{1,o} = 5.73,\; w_{2,o} = 5.83,\; b_o = -2.86
$$

**Activation Function (Sigmoid):**

$$
\sigma(z) = \frac{1}{1 + e^{-z}}
$$

---

# 2. Hidden Layer Calculations

## Neuron $H_1$

**Weighted Sum:**

$$
z_1 = (4.83)(1) + (4.83)(1) + (-2.82)
$$

$$
z_1 = 4.83 + 4.83 - 2.82 = 6.84
$$

**Activation:**

$$
a_1 = \sigma(6.84) = \frac{1}{1 + e^{-6.84}} \approx 0.9989
$$

---

## Neuron $H_2$

**Weighted Sum:**

$$
z_2 = (4.54)(1) + (4.60)(1) + (-2.74)
$$

$$
z_2 = 4.54 + 4.60 - 2.74 = 6.40
$$

**Activation:**

$$
a_2 = \sigma(6.40) = \frac{1}{1 + e^{-6.40}} \approx 0.9983
$$

---

# 3. Output Layer Calculation

## Output Neuron $O$

**Weighted Sum:**

$$
z_o = (5.73)(a_1) + (5.83)(a_2) + (-2.86)
$$

$$
z_o = (5.73)(0.9989) + (5.83)(0.9983) - 2.86
$$

$$
z_o \approx 5.724 + 5.820 - 2.86 = 8.684
$$

**Activation:**

$$
a_o = \sigma(8.684) = \frac{1}{1 + e^{-8.684}} \approx 0.9998
$$

---

# 4. Final Output

$$
\boxed{a_o \approx 0.9998}
$$

## Interpretation of Final Output

This value is extremely close to 1, indicating that the output neuron is highly activated. Since the sigmoid activation function maps inputs to a range between 0 and 1, values near 1 represent strong confidence in the positive class. This result occurs because both hidden layer neurons produced activations close to 1:
- $a_1 \approx 0.9989$
- $a_2 \approx 0.9983$

These high activations, when combined with the positive output weights ($w_{1,o} = 5.73$, $w_{2,o} = 5.83$), yield a large positive weighted sum ($z_o \approx 8.684$). Applying the sigmoid function to this large positive value pushes the final output very close to 1, but the expected value should have been 0. It is important to note that in the standard XOR truth table, $XOR(1,1) = 0$, since the XOR function outputs 1 only when the inputs differ and outputs 0 when they are the same.

Therefore, this result suggests that either the network is using a different label encoding or that the given weights and biases do not correctly represent the XOR function. In this case, both hidden neurons produce activations close to 1 for the input $(1,1)$, and the positive output weights amplify these values, resulting in a large positive weighted sum. Overall, the forward pass demonstrates how large positive weighted sums propagate through the sigmoid activation to produce outputs near 1.

```python
# XOR Neural Network Forward Pass
import numpy as np

# Sigmoid Activation Function
def sigmoid(z): return 1 / (1 + np.exp(-z))

# Input Vector
x1, x2 = 1, 1

# Weights and Biases
w11, w21 = 4.83, 4.83
b1 = -2.82

w12, w22 = 4.54, 4.60
b2 = -2.74

w1o, w2o = 5.73, 5.83
bo = -2.86

# Hidden Layer Calculations
z1 = w11*x1 + w21*x2 + b1
a1 = sigmoid(z1)

z2 = w12*x1 + w22*x2 + b2
a2 = sigmoid(z2)

# Output Layer Calculation
zo = w1o*a1 + w2o*a2 + bo
ao = sigmoid(zo)

# Print Results
print("== Hidden Neuron H\u2081 ==")
print("z\u2081 =", z1)
print("a\u2081 =", a1)
print("\n== Hidden Neuron H\u2082 ==")
print("z\u2082 =", z2)
print("a\u2082 =", a2)
print("\n== Output Neuron ==")
print("z\u2092 =", zo)
print("a\u2092 =", ao)
```
