A data set of sale prices is given as a Python list:

```python
prices = [299.0, 399.0, 1709.0, 349.0, 549.0, 239.0, 399.0, 399.9, 299.0, 898.5, 449.0]
```

Ex. 1. Calculate the average price.

```python
# Calculate the average price by dividing the sum of all prices by the number of prices.
average_price = sum(prices) / len(prices)

# Print the calculated average price.
print("Average Price:", average_price)
```

Ex. 2. The **median value** of a data set is the "middle" value. It is calculated as follows:
- Sort the data set. (You can use the [sort](https://www.w3schools.com/python/ref_list_sort.asp) method to sort a Python list.)
- If the size of the data set is odd, the median value is the value at index `size//2`.
- If the size of the data set is even, the median value is the average of the value at index `size//2 - 1` and `size//2`.

**Calculate the median price.**

```python
# Step 1: Sort the list of prices in ascending order.
sorted_prices = sorted(prices)

# Step 2: Calculate the median of the sorted prices.
n = len(sorted_prices)
if n % 2 == 1: # If the number of elements is odd, select the middle element as the median.
    median_price = sorted_prices[n // 2]
else:          # If the number of elements is even, calculate the median as the average of the two middle elements.
    median_price = (sorted_prices[n // 2 - 1] + sorted_prices[n // 2]) / 2
print("Median Price:", median_price)
```

Ex. 3. The **variance** is a descriptive statistic that measures how much a typical value in a data set differs from the average value. It is calculated as follows:
- Calculate the average value.
- For each value in the data set, calculate the square of the difference between the value and the average value. For example, if a value in the data set is 1.5 and the average value is 3.0, then calculate

$$
(1.5 - 3.0)^2 = 2.25.
$$

- The variance is the sum of these squares divided by `(length - 1)`.

**Calculate the variance of the sale prices.**

```python
def variance(prices):
    # This line calculates the average of the prices.
    average_price = sum(prices) / len(prices)

    # This variable stores the sum of squared differences from the mean.
    squared_differences_sum = 0

    # This loop adds each price's squared difference from the mean to the total.
    for price in prices:
        squared_differences_sum += (price - average_price) ** 2

    # This line calculates the variance by dividing by (length - 1).
    return squared_differences_sum / (len(prices) - 1)

# This line prints the variance of the sale prices.
print("Variance of the Sale Prices:", variance(prices))
```

Ex. 4 (Optional for undergradute students) Write a Python function `euclidean_distance(p1, p2)` that takes two Python lists representing points in n-dimensional space and calculate their difference. For example, if $p = [p_1, p_2, p_3, p_4]$ and $q = [q_1, q_2, q_3, q_4]$, then their distance is
$$
d = \sqrt{(p_1 - q_1)^2 + (p_2 - q_2)^2 + (p_3 - q_3)^2 + (p_4 - q_4)^2}
$$

```python
import math
def euclidean_distance(p1, p2):
    # This variable stores the sum of squared differences between coordinates.
    sum_of_squares = 0

    # This loop goes through each dimension of the points.
    for i in range(len(p1)):
        # Calculates the squared difference and adds it to the total.
        sum_of_squares += (p1[i] - p2[i]) ** 2

    # This line returns the square root of the sum of squared differences.
    return math.sqrt(sum_of_squares)

# Example Usage:
p = [1, 2, 3, 4]
q = [5, 6, 7, 8]

print("Euclidean Distance:", euclidean_distance(p, q))
```
